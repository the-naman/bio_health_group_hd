# One-time setup · bhg_dev
Creates the catalog, schemas and volumes for bhg_dev. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS bhg_dev COMMENT 'Bio Health Group, dev environment';
CREATE SCHEMA IF NOT EXISTS bhg_dev.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS bhg_dev.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS bhg_dev.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS bhg_dev.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS bhg_dev.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS bhg_dev.ops.refresh_in;
```
