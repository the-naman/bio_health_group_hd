# One-time setup · bhg_test
Creates the catalog, schemas and volumes for bhg_test. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS bhg_test COMMENT 'Bio Health Group, test environment';
CREATE SCHEMA IF NOT EXISTS bhg_test.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS bhg_test.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS bhg_test.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS bhg_test.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS bhg_test.bronze.landing;
CREATE VOLUME IF NOT EXISTS bhg_test.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS bhg_test.ops.refresh_in;
CREATE VOLUME IF NOT EXISTS bhg_test.ops.refresh_out;
```
