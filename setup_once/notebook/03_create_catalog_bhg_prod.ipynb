# One-time setup · bhg_prod
Creates the catalog, schemas and volumes for bhg_prod. Run once by hand. Not part of any job.

```sql
CREATE CATALOG IF NOT EXISTS bhg_prod COMMENT 'Bio Health Group, prod environment';
CREATE SCHEMA IF NOT EXISTS bhg_prod.bronze COMMENT 'raw data as landed';
CREATE SCHEMA IF NOT EXISTS bhg_prod.silver COMMENT 'clean, validated, SCD2 history';
CREATE SCHEMA IF NOT EXISTS bhg_prod.gold   COMMENT 'business-ready facts and dimensions';
CREATE SCHEMA IF NOT EXISTS bhg_prod.ops    COMMENT 'audit, watermarks, quality results';
CREATE VOLUME IF NOT EXISTS bhg_prod.ops.checkpoints;
CREATE VOLUME IF NOT EXISTS bhg_prod.ops.refresh_out;
```
